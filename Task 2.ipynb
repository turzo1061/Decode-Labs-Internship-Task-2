### Results Section (Template for IEEE Paper)

**Abstract:**

This section presents the comprehensive exploratory data analysis (EDA), data preprocessing pipeline, and initial model performance evaluation for a comparative machine learning approach aimed at stage-wise classification of pancreatic cancer. The objective was to predict the `Stage_at_Diagnosis` (Stage I, Stage II, Stage III, Stage IV) using various clinical features. The dataset, comprising 50,000 patient records and 24 features, underwent rigorous preprocessing and feature engineering to prepare it for robust model development. Baseline models were developed and evaluated both with and without dimensionality reduction using Principal Component Analysis (PCA) to assess its impact on predictive performance.

**I. Introduction:**

Pancreatic cancer remains one of the most aggressive malignancies with a grim prognosis, largely due to late-stage diagnosis. Early and accurate staging is critical for guiding treatment decisions and improving patient outcomes. This research focuses on developing a machine learning framework for the stage-wise classification of pancreatic cancer, leveraging a diverse set of clinical features. The study adheres to research-grade standards, including thorough data exploration, meticulous preprocessing, and comparative model evaluation, suitable for publication in leading scientific journals.

**II. Materials and Methods:**

**A. Dataset Description:**

The study utilized a dataset consisting of 50,000 records, each representing a unique patient, and 24 features. Key features included demographic information (e.g., Age, Gender, Country), lifestyle factors (e.g., Smoking History, Alcohol Consumption, Physical Activity Level, Diet Processed Food), medical history (e.g., Obesity, Diabetes, Chronic Pancreatitis, Family History, Hereditary Condition), presenting symptoms (e.g., Jaundice, Abdominal Discomfort, Back Pain, Weight Loss, Development of Type 2 Diabetes), and socio-economic indicators (e.g., Access to Healthcare, Urban vs Rural, Economic Status). The target variable, `Stage_at_Diagnosis`, is a multiclass categorical variable with four distinct classes: Stage I, Stage II, Stage III, and Stage IV.

**B. Exploratory Data Analysis (EDA):**

An extensive EDA was conducted to understand the dataset's characteristics. Initial observations revealed 4 duplicate records, which were subsequently handled. No missing values were identified across any features. Descriptive statistics provided insights into the central tendency, dispersion, and distribution of numerical and categorical variables. The target variable, `Stage_at_Diagnosis`, exhibited class imbalance, with Stage IV being the most prevalent (39.84%) and Stage I the least (9.87%). Univariate analysis included histograms and boxplots for numerical features and frequency plots for categorical features, highlighting their individual distributions. Bivariate analysis focused on the relationships between individual clinical factors and `Stage_at_Diagnosis`, utilizing count plots and boxplots. Multivariate analysis involved a correlation heatmap of numerical features, pairwise plots for selected features, and an assessment of multicollinearity using Variance Inflation Factor (VIF), identifying potential inter-dependencies among predictors.

**C. Data Leakage Assessment:**

Critical assessment for data leakage identified `Survival_Time_Months`, `Survival_Status`, and `Treatment_Type` as leakage-prone variables. These features, which represent outcomes post-diagnosis or post-treatment decisions, were removed from the dataset to ensure that the predictive model relies solely on information available at the time of diagnosis, preventing an inflated sense of model performance.

**D. Data Preprocessing Pipeline:**

1.  **Duplicate Handling:** The identified 4 duplicate records were removed from the dataset.
2.  **Missing Value Imputation:** No missing values were found, thus no imputation was required.
3.  **Categorical Feature Encoding:** All nominal categorical features (e.g., 'Country', 'Gender', 'Physical_Activity_Level') were transformed using One-Hot Encoding to convert them into a numerical format suitable for machine learning algorithms.
4.  **Target Label Encoding:** The ordinal `Stage_at_Diagnosis` target variable was encoded into numerical labels (0, 1, 2, 3) using LabelEncoder.
5.  **Feature Scaling:** Numerical features were scaled using `StandardScaler` (Z-score normalization) to standardize their ranges, preventing features with larger values from disproportionately influencing model training. A comparison with `MinMaxScaler` was performed to illustrate the differences in their transformation effects.
6.  **Class Imbalance Handling:** To address the inherent class imbalance in the `Stage_at_Diagnosis` variable, the Synthetic Minority Over-sampling Technique (SMOTE) was applied to the training data. This process synthesized new examples for minority classes, balancing the class distribution and improving the model's ability to learn from underrepresented stages.
7.  **Data Splitting:** The preprocessed and balanced dataset was split into training and testing sets using a stratified train-test split (80% training, 20% testing). Stratification ensured that the proportion of each cancer stage was maintained in both subsets, providing a representative evaluation.

**E. Feature Engineering:**

Various feature selection techniques were employed to identify the most salient features:

1.  **Chi-Square Test:** Applied to one-hot encoded categorical features to assess their independence from the target variable, identifying statistically significant predictors. The top 10 features selected by Chi-Square Test were: `['Country_United States', 'Country_India', 'Country_Germany', 'Gender_Male', 'Physical_Activity_Level_Medium', 'Physical_Activity_Level_Low', 'Diet_Processed_Food_Medium', 'Diet_Processed_Food_Low', 'Access_to_Healthcare_Medium', 'Urban_vs_Rural_Rural']`.
2.  **Mutual Information:** Used to quantify the dependency between each feature and the target variable, identifying both linear and non-linear relationships. The top 15 features selected by Mutual Information were: `['Age', 'Weight_Loss', 'Alcohol_Consumption', 'Abdominal_Discomfort', 'Back_Pain', 'Smoking_History', 'Gender_Male', 'Diabetes', 'Development_of_Type2_Diabetes', 'Obesity', 'Jaundice', 'Economic_Status_Middle', 'Access_to_Healthcare_Medium', 'Diet_Processed_Food_High', 'Diet_Processed_Food_Medium']`.
3.  **Recursive Feature Elimination (RFE):** A wrapper method that iteratively removes features based on their importance as determined by a base estimator (Logistic Regression), selecting a subset of features for optimal performance. The top 10 features selected by RFE were: `['Smoking_History', 'Diabetes', 'Chronic_Pancreatitis', 'Family_History', 'Hereditary_Condition', 'Jaundice', 'Abdominal_Discomfort', 'Back_Pain', 'Alcohol_Consumption', 'Urban_vs_Rural_Urban']`.
4.  **Random Forest Feature Importance:** Employed a Random Forest classifier to rank features based on their contribution to reducing impurity, providing a robust measure of predictive power. The top 15 features selected by Random Forest Importance were: `['Age', 'Weight_Loss', 'Alcohol_Consumption', 'Back_Pain', 'Smoking_History', 'Obesity', 'Country_United States', 'Abdominal_Discomfort', 'Diabetes', 'Development_of_Type2_Diabetes', 'Access_to_Healthcare_Medium', 'Jaundice', 'Economic_Status_Middle', 'Physical_Activity_Level_Medium', 'Physical_Activity_Level_Low']`.

**F. Dimensionality Reduction:**

Principal Component Analysis (PCA) was performed to reduce the dimensionality of the feature space while retaining maximum variance. An explained variance analysis revealed that approximately 29 principal components were required to capture 95% of the total variance in the dataset. Visualizations of 2D and 3D PCA projections were generated to observe the separability of the different cancer stages in a reduced feature space.

A Logistic Regression classifier was used as a baseline model to evaluate the impact of PCA. ## Data Leakage Assessment

Target leakage occurs when your training data includes information that would not be available at the time of prediction, i.e., information that 'leaks' from the target. For our task of predicting `Stage_at_Diagnosis`, any feature that is a direct consequence of or highly correlated with the diagnosis itself, or information that becomes available *after* diagnosis, could be a source of leakage.

Based on common medical datasets and the features provided, the following columns are potential candidates for target leakage:

*   `Survival_Time_Months`: This information is only known *after* diagnosis and subsequent observation. Including it would directly reveal aspects of disease progression post-diagnosis.
*   `Survival_Status`: Similar to `Survival_Time_Months`, this is an outcome *after* diagnosis and treatment.
*   `Treatment_Type`: The type of treatment is typically decided *after* diagnosis and staging. While it might correlate with stage, it's not a pre-diagnosis clinical feature useful for predicting the stage itself.

These features should ideally be excluded from the dataset for the stage classification model to prevent leakage and ensure the model's predictions are based purely on pre-diagnosis clinical features.

### Identify and Recommend Removal of Leakage-Prone Variables

```python
leakage_prone_variables = ['Survival_Time_Months', 'Survival_Status', 'Treatment_Type']

print(f"Identified leakage-prone variables: {leakage_prone_variables}")
print("These variables will be removed before training the classification model to prevent target leakage.")

# Create a copy of the dataframe for preprocessing to avoid modifying the original 'df'
df_preprocessed = df.copy()

# Remove leakage-prone variables
df_preprocessed = df_preprocessed.drop(columns=leakage_prone_variables)

print(f"Shape of dataframe after removing leakage-prone variables: {df_preprocessed.shape}")
```
